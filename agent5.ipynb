```python
from langgraph.graph import StateGraph, START, END
```

```python
from typing import Dict, TypedDict, List
import random
```

```python
class AgentState(TypedDict):
  name:str
  counter:int
  numbers:List[int]
```

```python
def greetingNode(state: AgentState) -> AgentState:
  """This is a Simple Greeting node"""
  print(f"welcome {state["name"]}")
```

```python
def randomNode(state: AgentState) -> AgentState:
  """This creates a random number"""
  state["numbers"].append(random.randint(0,100))
  state["counter"] += 1

  return state
```

```python
def shouldContinue(state:AgentState) -> AgentState:
  """Function to decide whats next"""
  if state["counter"] <= 5:
    print("Entering loop ", state["counter"])
    return "loop"
  else:
    return "exit"
```

```python
graph = StateGraph(AgentState)
graph.add_node("greetings", greetingNode)
graph.add_node("random",randomNode)
graph.add_edge("greetings","random")
graph.add_conditional_edges(
    "random",
    shouldContinue,
    {
        "loop":"random",
        "exit":END
    }
)

graph.set_entry_point("greetings")
app = graph.compile()
```

```python
res = app.invoke({"name":"Aditya" , "counter" : 0 , "numbers" : []})
print(res)
```
