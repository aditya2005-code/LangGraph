```python
from langgraph.graph import StateGraph, START, END
```

```python
from typing import Dict, TypedDict, List
```

```python
class AgentState(TypedDict):
  name:str
  result : str
  marks : List[int]
  overall:str


def greetingNode(state : AgentState) -> AgentState:
  """It is just a simple node that adds a greeting message to a state"""

  state["result"] = f"Hey {state['name']}, My total marks are {sum(state['marks'])}"
  return state

def gradeCheck(state : AgentState) -> AgentState:
  """It checks the student is pass or fail"""
  if sum(state["marks"]) > 150 :
    state["overall"] = f"congrats {state["name"]} you passed the exam"
  else:
    state["overall"] = f"Sorry, {state["name"]} try hard next time"

  return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("greeter",greetingNode)
graph.add_node("evaluator",gradeCheck)
graph.set_entry_point("greeter")
graph.add_edge("greeter","evaluator")
graph.set_finish_point("evaluator")

app = graph.compile()
```

```python
ans = app.invoke({"name" : "Aditya" , "marks" : [30,35,50,55]})
print(ans)
```
