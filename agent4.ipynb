```python
from langgraph.graph import StateGraph, START, END
```

```python
from typing import Dict, TypedDict, List
```

```python
class AgentState(TypedDict):
  name:str
  result : str
  marks : List[int]
  overall:str


```

```python

# 2. Define Nodes
def greetingNode(state: AgentState):
    print("Greeting Node Executed")

    return {
        "result": f"Hello {state['name']}, checking your marks..."
    }


def passNode(state: AgentState):
    print("Pass Node Executed")

    return {
        "result": f"Congratulations {state['name']}! You passed."
    }


def failNode(state: AgentState):
    print("Fail Node Executed")

    return {
        "result": f"Sorry {state['name']}, you failed."
    }
```

```python
# 3. Define Conditional Function
def gradeCheck(state: AgentState):

    if state["marks"] >= 40:
        return "pass"

    else:
        return "fail"
```

```python

# 4. Create Graph
graph = StateGraph(AgentState)

# Add Nodes
graph.add_node("greeter", greetingNode)
graph.add_node("pass", passNode)
graph.add_node("fail", failNode)

# Define Entry Point
graph.add_edge(START, "greeter")

# Add Conditional Edge
graph.add_conditional_edges(
    "greeter",
    gradeCheck,
    {
        "pass": "pass",
        "fail": "fail"
    }
)

# Define Finish Points
graph.add_edge("pass", END)
graph.add_edge("fail", END)

# 5. Compile Graph
app = graph.compile()
```

```python
# 6. Execute Graph
result = app.invoke({
    "name": "Aditya",
    "marks": 75,
    "result": ""
})

print(result)
```
